# Types of table or Volume

```sql
1. Managed table

create table alpha (id int)

table in catalog 
(data files) + delta log ( default storage)


DROP: you loose everything 

2. External Tables (Unmanaged Table)
create table alpha_ext (id int, name sting) location 's3:alpha'
table in catalog
(data files) + delta log ( s3:alpha )


DROp: only table is drop but metadata is untouched
```

```sql
create table dev.naval.alpha_ext (id int, name string) location 's3://jpmctraining/alpha/naval'
```

```sql
insert into dev.naval.alpha_ext values (1,'Alia')
```

```sql
undrop table dev.naval.alpha;
```

```sql
drop table dev.naval.alpha_ext;
```

```sql
create or replace table dev.naval.zeta as 
select * from delta.`s3://jpmctraining/alpha/naval`
```
