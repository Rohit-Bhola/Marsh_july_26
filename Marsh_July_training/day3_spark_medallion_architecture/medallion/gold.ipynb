```sql
create or replace view IDENTIFIER(concat(:catalog, '.', 'naval_gold.transactions_summary')) as  
select transaction_type, round(sum(amount)) as total_amount from IDENTIFIER(concat(:catalog, '.', 'naval_silver.transactions')) group by transaction_type order by total_amount desc
```

```sql
create or replace view IDENTIFIER(concat(:catalog, '.', 'naval_gold.amount_tier')) as 
select
  dev.naval_silver.classify_amount_tier(amount) as amount_tier,
  count(*) as transaction_count
from IDENTIFIER(concat(:catalog, '.', 'naval_silver.transactions'))
group by amount_tier
order by transaction_count desc
```
